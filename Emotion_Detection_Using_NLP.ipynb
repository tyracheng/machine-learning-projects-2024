This is a model that predicts emotions through whatever words the user types.



Paper: Pérez, J. M., Rajngewerc, M., Giudici, J. C., Furman, D. A., Luque, F., Alemany, L. A., & Martínez, M. V. (2023). pysentimiento: A Python Toolkit for Opinion Mining and Social NLP tasks. arXiv preprint arXiv:2106.09462.


```python
# Upgrade torchvision to be compatible with the installed torch 2.14.0
!pip install --upgrade torchvision

# Upgrade transformers to ensure it's up-to-date with torch 2.14.0
!pip install --upgrade transformers

# Reinstall pysentimiento to pick up the newly compatible dependencies
!pip install pysentimiento --force-reinstall

from pysentimiento import create_analyzer
```

```python
def emotion_detection(text):
  emotion = create_analyzer(task = 'emotion', lang = 'en')
  result = emotion.predict(text)
  return result.output
emotion_detection(input())
```

```python
hatespeech = create_analyzer(task = 'hate_speech', lang = 'en')
result = hatespeech.predict(input())
next(iter(result.probas.items()), None)
```

```python
def hatespeech_detection(text):
  hatespeech = create_analyzer(task = 'hate_speech', lang = 'en')
  result = hatespeech.predict(text)
  return next(iter(result.probas.items()), None)
hatespeech_detection(input())
```

```python
text = input()

emo = emotion_detection(text)
hate = hatespeech_detection(text)

print("Emotion:", emo)
print("Speech:", hate)
```
